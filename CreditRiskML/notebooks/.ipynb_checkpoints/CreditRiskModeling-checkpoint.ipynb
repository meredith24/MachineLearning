```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
%matplotlib inline
import sys
from pathlib import Path

ROOT = Path.cwd().parent
sys.path.append(str(ROOT))

from src.config import RAW_DATA_PATH
```

```python
df = pd.read_csv(RAW_DATA_PATH)
```

```python
df.head()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, OrdinalEncoder

X = df.drop("loan_status", axis=1)
y = df["loan_status"]

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, stratify=y, random_state=42)

cat_features = [col for col in df.columns if df[col].dtype == "str"]
num_features = [col for col in df.columns if df[col].dtype != "str"]

age_to_drop = df[df["person_age"] > 100].index
df = df.drop(age_to_drop).copy()

ordinal_features = ["loan_grade"]
nominal_features = ["loan_intent", "person_home_ownership"]
binary_features = ["cb_person_default_on_file"]








transformer = ColumnTransformer(
    transformers=[
    ("median_imputer", SimpleImputer(strategy="median"), ["person_emp_length"]),
    ("mean_imputer", SimpleImputer(strategy="mean"), ["loan_int_rate"]),
    ("onehot", OneHotEncoder(drop="first"), nominal_features),
    ("binary", OrdinalEncoder(categories=[["N", "Y"]]), binary_features),
    ("ordinal", OrdinalEncoder(categories=[["A", "B", "C", "D", "E", "F", "G"]]), ordinal_features)],
    remainder="passthrough"
)

X_train_preprocessed = transformer.fit_transform(X_train)
column_names = transformer.get_feature_names_out()
X_test_preprocessed = transformer.transform(X_test)
X_train_preprocessed = pd.DataFrame(data=X_train_preprocessed, columns=column_names)
X_test_preprocessed = pd.DataFrame(data=X_test_preprocessed, columns=column_names)
```

```python
X_train_preprocessed
```

```python
import math

def plot_all_histograms(df, title_prefix=""):
    num_cols = df.select_dtypes(exclude="O").columns
    n_cols = 3
    n_rows = math.ceil(len(num_cols) / n_cols)
    plt.figure(figsize=(3*n_cols, 2*n_rows))

    for i, col in enumerate(num_cols,1):
        plt.subplot(n_rows, n_cols, i)
        sns.histplot(df[col], bins=30, kde=True)
        plt.title(f"{title_prefix} {col}")
        plt.xlabel("")
        plt.ylabel("")

    plt.tight_layout()
    plt.show()
```

```python
plot_all_histograms(X_train_preprocessed)
```

```python
sns.histplot(data=y_train)
plt.show()
```

```python
from xgboost import XGBClassifier
from sklearn.metrics import classification_report, roc_auc_score, confusion_matrix, accuracy_score
```

```python
model = XGBClassifier(n_jobs=-1)
model.fit(X_train_preprocessed, y_train)
y_pred = model.predict(X_test_preprocessed)
print(classification_report(y_test, y_pred))
print(confusion_matrix(y_test, y_pred))
print(accuracy_score(y_test, y_pred))
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.model_selection import RandomizedSearchCV
params = {
    "n_estimators": [100,200,500],
    "max_depth": [3,7,15,30],
    "scale_pos_weight": [1, 1.5, (len(y_train[y_train==0] / len(y_train[y_train==1])))],
    "learning_rate": [0.01, 0.1, 0.9],
    "gamma": [3,5,7,None]
}

rscv = RandomizedSearchCV(estimator=XGBClassifier(), param_distributions=params, n_iter=30, scoring="f1", cv=3, random_state=42, n_jobs=-1)
```

```python
rscv.fit(X_train_preprocessed, y_train)
```

```python
y_pred = rscv.predict(X_test_preprocessed)
print(classification_report(y_test, y_pred))
print(confusion_matrix(y_test, y_pred))
print(accuracy_score(y_test, y_pred))
```

```python
Pt_x = PowerTransformer(method="yeo-johnson")
```

```python
pipeline = Pipeline(
    steps=[
        ("column_transformer", transformer),
        ("yeo-johnson", Pt_x),
        ("xgb_classifer", XGBClassifier() )
    ]
)
```

```python
pipeline.fit(X_train, y_train)
y_pred = pipeline.predict(X_test)
print(classification_report(y_test, y_pred))
print(confusion_matrix(y_test, y_pred))
print(accuracy_score(y_test, y_pred))
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler

pipeline_lr = Pipeline(
    steps=[
        ("column_transformer", transformer),
        ("std_scaler", StandardScaler()),
        ("yeo-johnson", Pt_x),
        ("logistic_regression", LogisticRegression(class_weight="balanced", random_state=42))
    ]
)
```

```python
pipeline_lr.fit(X_train, y_train)
y_pred = pipeline_lr.predict(X_test)
print(classification_report(y_test, y_pred))
print(confusion_matrix(y_test, y_pred))
print(accuracy_score(y_test, y_pred))
```

```python
pipeline_lr_unbalanced = Pipeline(
    steps=[
        ("column_transformer", transformer),
        ("std_scaler", StandardScaler()),
        ("yeo-johnson", Pt_x),
        ("logistic_regression", LogisticRegression(class_weight=None, random_state=42))
    ]
)
```

```python
pipeline_lr_unbalanced.fit(X_train, y_train)
y_pred = pipeline_lr_unbalanced.predict(X_test)
print(classification_report(y_test, y_pred))
print(confusion_matrix(y_test, y_pred))
print(accuracy_score(y_test, y_pred))
```

```python
from sklearn.metrics import roc_curve, roc_auc_score, precision_recall_curve, average_precision_score
```

```python
y_prob_default = rscv.predict_proba(X_test_preprocessed)[:,1] # first column -->  class 0(non-default), second column --> class 1(default)
```

```python
fpr, tpr, thresholds = roc_curve(y_test, y_prob_default)
roc_auc = roc_auc_score(y_test, y_prob_default)

plt.figure(figsize=(15,8))
plt.plot(fpr, tpr, label=f"AUC: {roc_auc: .3f}")
plt.plot([0,1], [0,1], "--")
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.legend()
plt.show()
```

```python
precision, recall, thresholds = precision_recall_curve(y_test, y_prob_default)
avg_precision_score = average_precision_score(y_test, y_prob_default)

plt.figure(figsize=(15,8))
plt.plot(precision, recall, label=f"AVG_PRE_SCORE: {avg_precision_score: .3f}")
plt.xlabel("Precision")
plt.ylabel("Recall")
plt.legend()
plt.show()
```

```python
from sklearn.model_selection import TunedThresholdClassifierCV
```

```python
classifier_tuned = TunedThresholdClassifierCV(estimator=XGBClassifier(), scoring="f1", thresholds=100, cv=5, n_jobs=-1, random_state=42)
```

```python
classifier_tuned.fit(X_train_preprocessed, y_train)
```

```python
classifier_tuned.best_threshold_
```

```python
classifier_tuned.best_score_
```

```python
classifier_tuned.estimator_
```

```python
y_pred = classifier_tuned.predict(X_test_preprocessed)
print(classification_report(y_test, y_pred))
print(confusion_matrix(y_test, y_pred))
print(accuracy_score(y_test, y_pred))
```

```python
classifier_tuned = TunedThresholdClassifierCV(estimator=XGBClassifier(), scoring="recall", thresholds=100, cv=5, n_jobs=-1, random_state=42)
classifier_tuned.fit(X_train_preprocessed, y_train)
print("best threshold: ", classifier_tuned.best_threshold_)
y_pred = classifier_tuned.predict(X_test_preprocessed)
print(classification_report(y_test, y_pred))
print(confusion_matrix(y_test, y_pred))
print(accuracy_score(y_test, y_pred))
```

```python
classifier_tuned = TunedThresholdClassifierCV(estimator=XGBClassifier(), scoring="precision", thresholds=100, cv=5, n_jobs=-1, random_state=42)
classifier_tuned.fit(X_train_preprocessed, y_train)
print("best threshold: ", classifier_tuned.best_threshold_)
y_pred = classifier_tuned.predict(X_test_preprocessed)
print(classification_report(y_test, y_pred))
print(confusion_matrix(y_test, y_pred))
print(accuracy_score(y_test, y_pred))
```

```python
classifier_tuned = TunedThresholdClassifierCV(estimator=XGBClassifier(), scoring="balanced_accuracy", thresholds=100, cv=5, n_jobs=-1, random_state=42)
classifier_tuned.fit(X_train_preprocessed, y_train)
print("best threshold: ", classifier_tuned.best_threshold_)
y_pred = classifier_tuned.predict(X_test_preprocessed)
print(classification_report(y_test, y_pred))
print(confusion_matrix(y_test, y_pred))
print(accuracy_score(y_test, y_pred))
```

```python
# varsayılanı 0.5 olan threshold ile elde edilen recall seviyesinin iyileştirilmesi için optimal olan 0.24 olan threshold seçildi.
```

```python
pipeline_final = Pipeline(
    steps=[
        ("column_transformer", transformer),
        ("yeo-johnson", Pt_x),
        ("tuned_classifier", classifier_tuned)
    ]
)
```

```python
pipeline_final.fit(X_train, y_train)
y_pred = pipeline_final.predict(X_test)
print(classification_report(y_test, y_pred))
print(confusion_matrix(y_test, y_pred))
print(accuracy_score(y_test, y_pred))
```
